```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Carga desde un archivo .xlsx sin indice
df= pd.read_excel('Citas_Digital.xlsx')
df
```

```python
df.info()
```

```python
#1. Filtro por filas
filtro1= df.iloc[0:673,: ]
filtro1.info()
```

```python
#2. Filtro por columnas
filtro1= filtro1.iloc[:,[0,1,2,4,5,6,7,8,9]].copy()
filtro1.info()
```

```python
filtro1.shape
```

```python
#3. Potencial de compra -> float
filtro1['Potencial de compra'] = filtro1['Potencial de compra'].astype('string')
```

```python
filtro1['Potencial de compra']=filtro1['Potencial de compra'].str.replace("Ventas ","0", regex=False)
```

```python
filtro1['Potencial de compra']=filtro1['Potencial de compra'].str.replace("%","", regex=False)
```

```python
filtro1['Potencial de compra']=filtro1['Potencial de compra'].str.replace("'","", regex=False)
```

```python
filtro1['Potencial de compra']=filtro1['Potencial de compra'].str.replace("&","", regex=False)
```

```python

filtro1['Potencial de compra']= filtro1['Potencial de compra'].astype(float)
```

```python
filtro1['Potencial de compra'].dtype
```

```python
filtro1.info()
```

```python
filtro1.to_csv("Citas_Digital_General.csv", index=False)
```
